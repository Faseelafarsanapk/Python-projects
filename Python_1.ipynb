```python
# String Concatenation:
string1 = "Hello "
string2 = "Faseela Farsana"

result = string1 + string2

print(result)

string3 = ", Welcome to Python Programming"

result = result + string3

print(result)
```

```python
#Write a Python program using the above concatenated string as input and performs the
# following tasks:

#String Slicing and Indexing:
string1 = "Hello "
string2 = "Faseela Farsana"
string3 = ", welcome to Python programming"

result = string1 + string2 + string3

# a. First character
print("First character:", result[0])

#b. Last charecter of result
print("Last charecter:", result[-1])

#c. First five charecters of result
print("First 5 charecters:", result[:5])

#d. Last 11 charecters of the result
print("Last 11 charecters:", result[-11:])

#e. Print the result in revers
print("revers of result:", result[::-1])

#f. Use slicing and print the word “Python” from the existing string.
print("Word Python:", result[24:30])
```

```python
# String Methods: Write a Python program that takes a string,
#strM = “Python beginner tutorial” and perform the following tasks:

strM = "Python beginner tutorial"

# a. Convert the sentence to uppercase
print("Uppercase:", strM.upper())

# b. Convert the sentence to lowercase
print("Lowercase:", strM.lower())

# c. Capitalize the sentence
print("Capitalize:", strM.capitalize())

# d. Count the total number of occurrences of character 't'
print("Occurrences of 't':", strM.count('t'))

# e. Replace all occurrences of "Python" with "Machine Learning"
print("After replacement:", strM.replace("Python", "Machine Learning"))
```

```python
#Tuples (Creation, Modification and Access) :
#Create 1st tuple with values -> (10, 20, 30), 2nd tuple with values -> (40, 50, 60):

# Created two tuples
tuple1 = (10, 20, 30)
tuple2 = (40, 50, 60)

# a. Concatenate the two tuples
t_combine = tuple1 + tuple2
print("Combined tuple:", t_combine)

# b. Repeat the elements of t_combine 3 times
print("Repeated tuple:", t_combine * 3)

# c. Access the 3rd element from t_combine
print("3rd element:", t_combine[2])

# d. Access the first three elements
print("First three elements:", t_combine[:3])

# e. Access the last three elements
print("Last three elements:", t_combine[-3:])
```
